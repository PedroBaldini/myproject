```python
import duckdb
```

```python
con = duckdb.connect(database='dados_duck.db', read_only=False)
```

```python
# Verificando os dados da camada Silver
df = con.execute("SELECT * FROM produtos").fetchdf()
df.head(10)
```

```python
# Criando uma tabela dimensao na camada Gold
df2 = df.drop(columns=['id_categoria', 'id_fornecedor'])
df2.head(10)
```

```python
con.execute("""
    DROP TABLE dim_produtos
""")
```

```python
con.execute("""
    CREATE TABLE IF NOT EXISTS dim_produtos(
            id_produto BIGINT,
            nm_produto TEXT,
            vl_produto FLOAT
            )
""")
```

```python
con.execute("INSERT INTO dim_produtos SELECT * FROM df2")
```

```python
df_dim = con.execute("SELECT * FROM dim_produtos").fetchdf()
df_dim.head(10)
```
